```python
from google.colab import files
import subprocess
import os
import glob

print("Seleziona il file video da caricare dal tuo computer (es. vacanza3.mp4):")
up = files.upload()
if len(up) > 0:
    VIDEO = list(up.keys())[0]
    print(f"File '{VIDEO}' caricato con successo!")

    # Creazione cartella frames ed estrazione fotogrammi
    os.makedirs('frames', exist_ok=True)
    print("Estrazione dei fotogrammi con ffmpeg...")
    subprocess.run(['ffmpeg', '-loglevel', 'error', '-y', '-i', VIDEO, '-q:v', '2', '-start_number', '0', 'frames/%05d.jpg'], check=True)

    # Ottenimento FPS e numero di fotogrammi
    r = subprocess.check_output(['ffprobe', '-v', 'error', '-select_streams', 'v:0', '-show_entries', 'stream=r_frame_rate', '-of', 'csv=p=0', VIDEO]).decode().strip()
    a, b = r.split('/')
    FPS = float(a) / float(b)
    N = len(glob.glob('frames/*.jpg'))
    print(f"Estratti correttamente {N} fotogrammi a {FPS:.2f} FPS.")
else:
    print("Nessun file caricato. Carica un video per procedere!")
```

```python
import cv2
import matplotlib.pyplot as plt

def show_grid(frame_idx, step=100):
    # Carica l'immagine e mostra la griglia per identificare le coordinate x, y
    img_path = f'frames/{frame_idx:05d}.jpg'
    if not os.path.exists(img_path):
        print(f"Fotogramma {frame_idx} non trovato!")
        return
    im = cv2.cvtColor(cv2.imread(img_path), cv2.COLOR_BGR2RGB)
    h, w = im.shape[:2]
    plt.figure(figsize=(10, 10))
    plt.imshow(im)
    plt.xticks(range(0, w, step))
    plt.yticks(range(0, h, step))
    plt.grid(alpha=0.5, color='yellow')
    plt.title(f"Fotogramma {frame_idx} con griglia di puntamento")
    plt.show()

# Mostriamo il fotogramma iniziale (0) per scegliere i punti
show_grid(0)
```

```python
import torch
import numpy as np
from sam2.build_sam import build_sam2_video_predictor

# Carichiamo il modello SAM 2 usando la GPU
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Utilizzo dispositivo: {device}")

model_cfg = "configs/sam2.1/sam2.1_hiera_b+.yaml"
predictor = build_sam2_video_predictor(model_cfg, checkpoint_path, device=device)
inference_state = predictor.init_state(video_path="frames")

# --- CONFIGURA QUI I TUOI PUNTI DA TRACCIARE ---
# Inserisci le coordinate [x, y] ricavate dalla griglia del fotogramma sovrastante.
# 'pos' indica i punti interni all'oggetto da tenere, 'neg' i punti dello sfondo da escludere.
PROMPTS = {
    0: {
        'pos': [
            [250, 450] # Esempio di punto da tenere (sostituisci con le tue coordinate!)
        ],
        'neg': [
            [10, 10]    # Esempio di punto da escludere (opzionale)
        ]
    }
}

# Funzione per registrare i punti nel motore di tracking di SAM 2
predictor.reset_state(inference_state)
for frame_idx, pt in PROMPTS.items():
    points = np.array(pt['pos'] + pt.get('neg', []), dtype=np.float32)
    labels = np.array([1]*len(pt['pos']) + [0]*len(pt.get('neg', [])), dtype=np.int32)
    _, out_obj_ids, out_mask_logits = predictor.add_new_points_or_box(
        inference_state=inference_state,
        frame_idx=frame_idx,
        obj_id=1,
        points=points,
        labels=labels
    )

print("Modello configurato e punti registrati. Siamo pronti per elaborare il video!")
```

```python
import numpy as np
import os
import cv2
import subprocess
import torch
from google.colab import files
from sam2.build_sam import build_sam2_video_predictor

# Verifica e inizializzazione sicura del modello se non presente nel kernel
if 'predictor' not in globals() or 'inference_state' not in globals():
    print("Inizializzazione del modello predictor e dello stato...")
    device = "cuda" if torch.cuda.is_available() else "cpu"
    checkpoint_path = 'checkpoints/sam2.1_hiera_base_plus.pt'
    model_cfg = "configs/sam2.1/sam2.1_hiera_b+.yaml"
    predictor = build_sam2_video_predictor(model_cfg, checkpoint_path, device=device)
    inference_state = predictor.init_state(video_path="frames")

    # Usa i PROMPTS definiti o inserisce un valore predefinito sicuro
    pts = PROMPTS if 'PROMPTS' in globals() else {0: {'pos': [[250, 450]]}}
    predictor.reset_state(inference_state)
    for frame_idx, pt in pts.items():
        points = np.array(pt['pos'] + pt.get('neg', []), dtype=np.float32)
        labels = np.array([1]*len(pt['pos']) + [0]*len(pt.get('neg', [])), dtype=np.int32)
        predictor.add_new_points_or_box(inference_state=inference_state, frame_idx=frame_idx, obj_id=1, points=points, labels=labels)

# Creazione cartella per le maschere di output
os.makedirs('masks', exist_ok=True)

print("Avvio del tracciamento di SAM 2 su tutto il video...")
# Propagazione del tracciamento nel video
video_segments = {}
for out_frame_idx, out_obj_ids, out_mask_logits in predictor.propagate_in_video(inference_state):
    video_segments[out_frame_idx] = {
        out_obj_id: (out_mask_logits[i] > 0.0).cpu().numpy()
        for i, out_obj_id in enumerate(out_obj_ids)
    }

# Salva le maschere come immagini PNG binarie
print("Salvataggio delle maschere binarie...")
for frame_idx in range(N):
    mask = np.zeros((cv2.imread(f'frames/{0:05d}.jpg').shape[:2]), dtype=np.uint8)
    if frame_idx in video_segments:
        for obj_id, obj_mask in video_segments[frame_idx].items():
            mask = np.maximum(mask, (obj_mask[0] * 255).astype(np.uint8))
    cv2.imwrite(f'masks/{frame_idx:05d}.png', mask)

# 8. Esporta il video con trasparenza
print("Applicazione della trasparenza (Feathering)...")
FEATHER = 1.0
os.makedirs('rgba', exist_ok=True)
for i in range(N):
    img = cv2.imread(f'frames/{i:05d}.jpg')
    m = cv2.imread(f'masks/{i:05d}.png', 0)
    if FEATHER > 0:
        m = cv2.GaussianBlur(m, (0, 0), FEATHER)
    cv2.imwrite(f'rgba/{i:05d}.png', np.dstack([img, m]))

print("Generazione dei file video (webm e mov) con ffmpeg...")
# Genera WebM trasparente
subprocess.run([
    'ffmpeg', '-y', '-loglevel', 'error', '-framerate', str(FPS),
    '-i', 'rgba/%05d.png', '-i', VIDEO,
    '-map', '0:v', '-map', '1:a?', '-c:v', 'libvpx-vp9',
    '-pix_fmt', 'yuva420p', '-b:v', '0', '-crf', '22', 'video_trasparente.webm'
], check=True)

# Genera ProRes 4444 MOV trasparente
subprocess.run([
    'ffmpeg', '-y', '-loglevel', 'error', '-framerate', str(FPS),
    '-i', 'rgba/%05d.png', '-i', VIDEO,
    '-map', '0:v', '-map', '1:a?', '-c:v', 'prores_ks',
    '-profile:v', '4', '-vendor', 'apl0', '-pix_fmt', 'yuva444p10le', 'video_trasparente.mov'
], check=True)

print("Video generati con successo! Avvio download automatico...")
files.download('video_trasparente.webm')
files.download('video_trasparente.mov')
```
